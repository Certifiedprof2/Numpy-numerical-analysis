### Question 1: Business Analytics — Sales Performance Calculator

```python
# The management team wants to understand the week's performance.
# The business recorded the following daily sales for one week:

import numpy as np

daily_sales = np.array([125000, 150000, 175000, 140000, 190000, 210000, 160000])
print()
print("Daily Sales:", daily_sales)
print(type(daily_sales))
print(daily_sales.shape)

# To calculate total sales 
total_sales = np.sum(daily_sales)
print("Total sales:", total_sales)

# To the Calculate Average Daily Sales
average_sales = np.mean(daily_sales)
print()
print("Average Sales:", average_sales)

# Apply a Sales Adjustment
adjusted_sales = daily_sales * 1.10
print()
print("Adjusted Sales:", adjusted_sales * 10)

# Task 5  Compare Original and Adjusted Sales
daily_sales = np.array([125000, 150000, 175000, 140000, 190000, 210000, 160000]) # original sales
adjusted_sales = daily_sales * 1.10  # Adjusted sales
print()
print("Original Sales:", daily_sales)
print("Adjusted Sales:", adjusted_sales)


```

### Looking at the above result the question is to know What does the average daily sales figure tell the business that the total sales figure does not?

#### Total sales: 1150000
#### Average Sales: 164285.71

### Explanation

#### The total sales figure shows the business that it generated ₦1,150,000 in sales over the 7-day period.

#### The average daily sales figure shows the business that it generated approximately ₦164,285.71 per day on average.

#### The average is useful because it helps the business understand its typical daily sales performance, while the total only shows the overall sales for the entire period.

## Exercise 2: Student Performance Analysis

### The program coordinator wants to understand the overall performance of the students and how individual scores differ from the group average.


```python
# Task 1 — Create and Inspect the Data
import numpy as np

scores = np.array([62, 75, 81, 69, 88, 94, 73, 85, 77, 91]) 
print("Score:", scores)
print(type(scores))
print(scores.shape)
print()

#Task 2 — Calculate the Mean Score
mean_score = np.mean(scores)
print("Mean Score:", mean_score)
print()

# Task 3 — Calculate Each Student's Deviation
deviations = scores - mean_score
print("Deviations:", deviations)
print()

#Task 5 — Square the Deviations
square_deviations = deviations ** 2
print("Square Deviation:", square_deviations)
print()
# # Squaring the deviations changes negative values into positive values
# because each deviation is multiplied by itself.

# Task 6 — Calculate the Sum of Squared Deviations
sum_squared_deviations = np.sum(square_deviations)
print("Sum squared deviations:", sum_squared_deviations)
print()
```

## Task 7 — Connect the Steps to Standard Deviation

### The stages involved in calculating standard deviation are:

####  Calculate the mean
#### Calculate the deviations from the mean
#### Square the deviations
#### Sum the squared deviations
#### Divide by the appropriate number of values
#### Take the square root

#### Standard deviation helps us understand how spread out the students' scores are from the mean score.
#### A smaller standard deviation means the scores are closer to the mean, 
#### while a larger standard deviation means the scores are more spread out.

### Engineering & Scientific Computing — Trigonometric Series and Convergence

```python
# Task 1 — Define the Angle
theta = 30   #This represents an angle in degrees.
theta_radian = np.radians(theta)

print("Theta Radian:", theta_radian)
print()

# Task 2: Task 2 — Create the Sequence of Terms
k = np.arange(1, 100, 1)
print(k)
print()

#Task 3 — Calculate the Trigonometric Component
sine_value = np.sin(theta_radian)
print("sine Value:", sine_value)
print()

# Task 4 — Build the Series Terms
series_terms = np.sin(theta_radian) / k
print(series_terms)
print()

# Task 5 — Calculate the Sum
sum_total = np.sum(series_terms)
print("Total Sum:", sum_total)
print()

# Task 6 — Increase the Number of Terms
k = np.arange(1, 100, 1)
series_terms = np.sin(theta_radian) / k
sum_total = np.sum(series_terms)

print("sum 99 Total:", sum_total)

k = np.arange(1, 1000, 1)
series_terms = np.sin(theta_radian) / k
sum_total = np.sum(series_terms)

print("sum 999 Total:", sum_total)


k = np.arange(1, 10000, 1)
series_terms = np.sin(theta_radian) / k
sum_total = np.sum(series_terms)

print("sum 9,999 Total:", sum_total)

```

### Task 7 — Investigate Convergence

### | Number of Terms | Series Result |
### | 99              | 2.5887 |
### | 999             | 3.7422 |
### | 9,999           | 4.8938|

#### Observation

#### As the number of terms increases, the series result also increases.
#### However, the increase becomes smaller as more terms are added.
#### This shows that the series is gradually approaching a stable value.

### Task 8 — Explain Convergence

###Convergence means that a result gets closer and closer to a particular value as more terms are added to a series.

#### What happened when I used fewer terms? 
##### When I used fewer terms, the series result was a smaller approximation because only a limited number of terms were included.

#### What happened when I increased the number of terms?
##### When I increased the number of terms, the series result increased and became a better approximation of the series.

#### Why does increasing the number of terms change the approximation?* 
##### Increasing the number of terms means that more values are added to the series. These additional terms contribute to the total, so the result changes.

#### What does it mean when the result gets closer to a limiting value?**  
##### It means that the series is converging. As more terms are added, the result gets closer to a particular value, called the limiting value.

### Monthly Business Expense Analysis Using NumPy

## Scenario
### A small business wants to analyze its monthly expenses using NumPy.
### The business records expenses for January to June and wants to understand
### its total expenses, average monthly expense, and how expenses change from month to month.

## Objective
### The objective of this experiment is to use NumPy to calculate and analyze
### monthly business expenses and understand how numerical operations can helpwith simple financial analysis.

```python
import numpy as np

# Monthly business expenses from January to June
expenses = np.array([120000, 135000, 128000, 150000, 142000, 160000])

print("Monthly Expenses:", expenses)
print()

# To calculate total expenses
total_expenses = np.sum(expenses)

print("Total Expenses:", total_expenses)
print()

average_expense = np.mean(expenses)

print("Average Monthly Expense:", average_expense)   
```

## Interpretation

#### The analysis shows the total amount spent by the business over the six-monthperiod and the average monthly expense.
#### The monthly expense changes show how spending increased or decreased from onemonth to the next. This can help the business identify months with higher expenses and monitor changes in its spending pattern.
